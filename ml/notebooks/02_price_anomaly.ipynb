# Held-out price anomaly evaluation

Run `python -m ml.evaluate --database-url $env:TRACEFOOD_OWNER_URL` from the project root to retrain and evaluate. The model is fitted to the oldest 70 percent of dates. Threshold selection uses the next 15 percent, and the rows below come from the final 15 percent.

```python
from pathlib import Path
import json

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / 'docs' / 'results.json').exists())
results = json.loads((root / 'docs' / 'results.json').read_text(encoding='utf-8'))
import pandas as pd

price = results['price_anomaly']
display(pd.Series({
    'training_end': results['training_end'],
    'validation_end': results['validation_end'],
    'training_rows': price['training_rows'],
    'validation_rows': price['validation_rows'],
    'held_out_rows': price['test_rows'],
    'validation_percentile_cutoff': price['threshold'],
}))
display(pd.DataFrame(
    [{'label': label, 'method': method, **metrics}
     for label, methods in price['held_out'].items()
     for method, metrics in methods.items()]
).set_index(['label', 'method']))
```

Per-label comparisons exclude observations injected with the other price-anomaly label. The overall comparison includes both labels. Results apply to synthetic data.
