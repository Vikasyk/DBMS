# Exploratory analysis of synthetic TraceFood records

Generate the seed-42 data first if the CSV files are absent. This analysis reads labels from their evaluation-only file.

```python
from pathlib import Path
import json

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / 'docs' / 'results.json').exists())
results = json.loads((root / 'docs' / 'results.json').read_text(encoding='utf-8'))
import pandas as pd

data = root / 'data' / 'synthetic'
batches = pd.read_csv(data / 'batch.csv')
prices = pd.read_csv(data / 'price_record.csv', parse_dates=['price_date'])
labels = pd.read_csv(data / 'labels.csv')
display(batches.groupby(['product_name', 'unit']).size().rename('batches'))
display(labels.groupby(['entity', 'label']).size().rename('planted cases'))
display(prices.groupby('supply_chain_stage').selling_price.agg(['count', 'median', 'min', 'max']))
```

These values describe the generated data. Price comparisons in the application use matching product, unit, stage, and market.
