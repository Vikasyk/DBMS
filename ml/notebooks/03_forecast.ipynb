# Rolling evaluation of the Bengaluru Tomato series

The target is mean weekly retail price in INR per kg. All candidate methods use the same rolling origins with forecasts four weeks ahead. Select the method with the lowest measured MAE.

```python
from pathlib import Path
import json

root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / 'docs' / 'results.json').exists())
results = json.loads((root / 'docs' / 'results.json').read_text(encoding='utf-8'))
import pandas as pd

forecast = results['forecast_tomato_bengaluru']
display(pd.Series({'consecutive_weeks': forecast['consecutive_weeks'],
                   'selected_model': forecast['model']}))
display(pd.DataFrame(forecast['metrics']).set_index('model').sort_values('mae'))
display(pd.DataFrame(forecast['forecast']).rename(columns={'price': 'INR per kg'}).set_index('week'))
```

The model comparison uses one synthetic market and product. A completed retail series with fewer than 52 consecutive weeks receives no forecast.
